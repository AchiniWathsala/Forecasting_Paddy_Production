```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

```

```python
data = pd.read_excel("./data/paddy_production_data.xlsx", header=None)

data.head()
```

```python
# name the columns
data.columns = ["Year", "Season", "Sown_Extent_Major", "Sown_Extent_Minor", "Sown_Extent_Rainfall", "Sown_Extent_Total",
                 "Harvested_Extent_Major", "Harvested_Extent_Minor", "Harvested_Extent_Rainfall", "Harvested_Extent_Total",
                "Avg_Yield_Major", "Avg_Yield_Minor", "Avg_Yield_Rainfall", "Avg_Yield_Total",                    
                "Production"]
data.head()
```

```python
# remove first two rows 
data = data.iloc[2:,]
data
```

```python
# Change the data type of the variables
data = data.astype({
    # "Year" : ,
    # "Season" : ,
    "Sown_Extent_Major" : "int", 
    "Sown_Extent_Minor" : "int",
    "Sown_Extent_Rainfall" : "int",
    "Sown_Extent_Total" : "int",
    "Harvested_Extent_Major" : "int",
    "Harvested_Extent_Minor" : "int",
    "Harvested_Extent_Rainfall" : "int",
    "Harvested_Extent_Total" : "int",
    "Avg_Yield_Major" : "int",
    "Avg_Yield_Minor" : "int",
    "Avg_Yield_Rainfall" : "int",
    "Avg_Yield_Total" : "int",                  
    "Production" : "float"
})
```

```python
data.info()
```

```python
def create_date(row):
    if row['Season'].lower() == "yala":
        return pd.Timestamp(year = row['Year'], month=4, day=1)
    elif row['Season'].lower() == "maha":
        return pd.Timestamp(year = row['Year']-1, month=9, day=1)
    
data['Date'] = data.apply(create_date, axis=1)

# sort for time series
data = data.sort_values('Date')

# set as index (important for forecasting)
#data = data.set_index('Date')

data.head()
```

```python
data1 = pd.read_excel("./data/paddy_production_data.xlsx", sheet_name= "Sheet2", header=None)
data1
```

```python
type(data1)
```

```python
data1.columns = [
    "Year", "Season",
    # Method of Sowing
    "Broadcasting","Transplant_Rows","Transplant_NotRows","Parachute","Machine","Other(sowing)","Row_Seeded",
    # Application of fertilizer
    "Chemical_only","Organic_only","Both","None",
    # Method of weeding
    "Hand_Weeding","Weedicide","Water","No_Weeding",
    # Use of insecticides
    "Insect_Used", "Insect_NotUsed",
    # Use of fungicide
    "Fung_Used","Fung_NotUsed",
    # Type of Seed variety
    "Short_Grain_Red","Long_Grain_Red","Short_Grain_White","Long_Grain_White","Other(seed_variety)",
    # Variety of Seed
    "New_Improved","Old_Improved","Traditional",
    
    
]
data1.head()
```

```python
data1 = data1.iloc[2:]
data1
```

```python
excleude_col = ["Year","Season"]

col_to_convert = data1.columns.difference(excleude_col)



data1[col_to_convert] = data1[col_to_convert].astype(float)

```

```python
data1.info()
```

```python
data1["Date"] = data1.apply(create_date,  axis=1)

data1 = data1.sort_values('Date')
data1
```

```python
data1[col_to_convert] = data1[col_to_convert].round(2)
```

```python
data1.head()
```

```python
df = pd.merge(data, data1, on='Date', how = 'left')
df
```

```python
df.isnull().sum()
```

```python
df = df.fillna(0)
df.head()
```

```python
df = df.drop(["Year_y","Season_y"],axis=1)
df = df.rename(columns={"Year_x": "Year", 
                        "Season_x": "Season"})
```

```python
df = df.set_index("Date")
```

```python
with pd.ExcelWriter("df.xlsx") as writer:
    df.to_excel(writer, sheet_name="df")
    data.to_excel(writer, sheet_name="data")
    data1.to_excel(writer, sheet_name="data1")
```

```python
import seaborn as sns
df3 = df2.copy()
df3.index = df3.index.year
df3 = df3.sort_index(ascending=False)
plt.figure(figsize=(10,8))
sns.heatmap(df3[['Broadcasting',
       'Transplant_Rows', 'Transplant_NotRows', 'Parachute', 'Machine',
       'Other(sowing)']], cmap="YlGnBu",linewidths=0.5,      # thickness of gap
       linecolor='gray',
       vmin=0,   # minimum %
       vmax=100,   # maximum %
       cbar_kws={'format': '%.0f%%'}).set_xticklabels(["Broadcasting","Transplanted in Rows","Transplanted not in Rows", "Parachute Mehtod", "Planting by Machinery","Other"], rotation =60)


plt.title("Year-wise Heatmap of Sowing Methods")
plt.xlabel("Method of Sowing")
plt.ylabel("Year")
plt.show()
```
