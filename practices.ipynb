```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
import matplotlib.dates as mdates
 
```

```python
data = pd.read_excel("./data/practices.xlsx", header=None)

data.head()
```

```python
data.columns = [
    "Year", "Season",
    # Method of Sowing
    "Broadcasting","Transplant_Rows","Transplant_NotRows","Parachute","Machine","Other(sowing)",
    # Application of fertilizer
    "Chemical_only","Organic_only","Both","None",
    # Method of weeding
    "Hand_Weeding","Weedicide","Water","No_Weeding"
    
    
]
data.head()
```

```python
data = data.iloc[2:]
data
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
excleude_col = ["Year","Season"]

col_to_convert = data.columns.difference(excleude_col)



data[col_to_convert] = data[col_to_convert].astype(int)

```

```python
data.info()
```

```python
def create_date(row):
    if row['Season'].lower() == "yala":
        return pd.Timestamp(year = row['Year'], month=4, day=1)
    elif row['Season'].lower() == "maha":
        return pd.Timestamp(year = row['Year']-1, month=9, day=1)
    
data["Date"] = data.apply(create_date,  axis=1)

data = data.sort_values('Date')
data

data = data.set_index("Date")
```

```python
data.shape
```

```python
with pd.ExcelWriter("df_practices.xlsx") as writer:
    data.to_excel(writer, sheet_name="df")
```
